# Skill Extraction Upgrade
This notebook documents and validates the improved resume/job-description skill extraction logic in `utils/parser.py`.

The notebook includes sample text cases, parser output checks, and a quick confirmation that the updated matching logic works as expected.


```python
from utils.parser import extract_skills, extract_job_description_skills, match_job_description

resume_text = '''
John Doe

Summary
Experienced software engineer with expertise in Python, Flask, Docker, SQL, and Git.

Skills
- Python
- Flask
- SQL
- Git
- Docker
- REST APIs
'''

job_description = '''
We are looking for a Python Developer.

Responsibilities:
- Develop web applications.
- Collaborate with engineering teams.

Skills:
Python, Flask, SQL, Git, Docker, CI/CD
'''

print('Resume Skills:')
print(extract_skills(resume_text))
print('\nJob Description Skills:')
print(extract_job_description_skills(job_description))
print('\nMatch Result:')
matched, missing, percentage = match_job_description(extract_skills(resume_text), job_description)
print('Matched:', matched)
print('Missing:', missing)
print('Match Percentage:', percentage)
```
